```python
# Yana ham takommilashgani ekan, sababi o'rtada internet uzulib qolsa dastur birdaniga to'xtab qolmas ekan

import time 
import pandas as pd
import random 
import requests 
from requests.exceptions import RequestException

def get_movie_links(page, retries =3):  # bu 3 martagacha qayta urinadi.
    url = f"https://www.themoviedb.org/movie?page={page}"
    for attempt in range(retries):
        try:
            response = requests.get(
                url, 
                headers=headers,
                timeout = 20
            )
            response.raise_for_status()
            soup = BeautifulSoup(response.text, "html.parser")
            links = set()

            for a in soup.find_all("a", href=True):
                href = a["href"]
                if re.match(r"^/movie/\d+", href):
                    links.add(BASE_URL + href)
            return list(links)
        except RequestException as e:
            print(f"Page {page} | Attempt {attempt + 1} / {retries} failed!")
            print(e)
            time.sleep(random.uniform(2, 5))
    print(f"Skippig page {page}")
    return []
```

```python
# Hozir demak noldan funksiya yoziladi. Shu orqali umumiy bitta DataFrame shkliga olib kelish uchun

import requests
from bs4 import BeautifulSoup
import re

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept-Language":"en-US, en; q = 0.9"
}
url = "https://www.themoviedb.org/movie/687163-project-hail-mary?language=en-US"  # istalgan birini sinab ko'rish uchun

# $ belgini tozalaymiz 
def clean_money(value):
    if value is None:
        return None
    if isinstance (value, int):
        return value
    value = value.strip()
    if value == "-" or value == "":
        return None
    
    return int(
        value.replace("$", "").replace(",", "").split(".")[0] # $ olib tashlab va , o'rniga . bilan berish
    )


def clean_runtime(runtime):
        if runtime is None:
            return None

        hours = 0
        minutes = 0

        h = re.search(r"(\d+)h", runtime) # Shunga e'tibor beish kerak
        m = re.search(r"(\d+)m", runtime) # Shunga e'tibor berish kerak

        if h:
            hours = int(h.group(1))

        if m:
            minutes = int(m.group(1))

        return hours * 60 + minutes

def parse_movie(url, retries=3):

    for attempt in range(retries):
        try:
            response = requests.get(
                url, 
                headers=headers, 
                timeout = 20
            )
            response.raise_for_status()

            soup = BeautifulSoup(response.text, "html.parser")
            break 

        except RequestException as e:

            print(f"Movie request failed ({attempt + 1}/{retries})")
            print(e)
            time.sleep(random.uniform(2, 5))
    else:
        return None


    # movie_idlarni olish 
    match = re.search(r'/movie/(\d+)', url)
    movie_id = int(match.group(1)) if match else None

    # title 
    title = soup.find("h2").get_text(" ", strip=True)
    title = re.sub(r"\s*\(\d{4}\)$", "", title)

    # release date
    release = soup.find("span", {"class": "release"})

    release_date = None
    release_country = None

    if release:
        release = release.get_text(strip=True)

        release_date = release.split(" (")[0]
        release_country = release.split("(")[1].replace(")", "")

        from datetime import datetime
        release_date = datetime.strptime(
            release_date,
            "%m/%d/%Y"
        ).strftime("%Y-%m-%d")

    # runtime
    runtime = soup.find("span", class_="runtime")
    runtime = runtime.get_text(strip=True) if runtime else None

    runtime = clean_runtime(runtime)

    # genres
    genres = soup.find("span", class_="genres")
    genres = ", ".join(
        [g.get_text(strip=True) for g in genres.find_all("a")]
    ) if genres else None

    # user score 
    score = soup.find("div", class_="user_score_chart")
    score = float(score["data-percent"]) / 10 if score else None

    # overview
    overview = soup.find("div", class_="overview")
    overview = overview.get_text(strip=True) if overview else None

    # Director 
    directors = []
    writers = []
    people = soup.find_all("li", class_="profile")
    for person in people:
        p = person.find_all("p")
        if len(p)>=2:
            name = p[0].get_text(strip=True)
            role = p[1].get_text(strip=True)
            print(name, "-->", role)
            if "Director" in role:
                directors.append(name)
            if "Writer" in role or "Screenplay" in role or "Novel" in role:
                writers.append(name)
    directors = ", ".join(directors) if directors else None
    writers = ", ".join(writers) if writers else None

    #Bizga kerak bo'lgan ustunlar masalan, budget, revenue,status, original_language. va production_companies'larni olamiz
    facts = soup.find("section", class_="facts")
    facts_data = {}

    if facts:

        for p in facts.find_all("p"):
            strong = p.find("strong")

            if strong:
                key = strong.get_text(strip=True)
                value = p.get_text(" ", strip=True)
                value = value.replace(key, "").strip()
                facts_data[key] = value

    facts_clean = {
        "status":facts_data.get("Status"),
        "original_language": facts_data.get("Original Language"),
        "budget": clean_money(facts_data.get("Budget")),
        "revenue": clean_money(facts_data.get("Revenue"))
    }

    return {
        "movie_id": movie_id,
        "url":url,
        "title":title,
        "runtime":runtime,
        "release_date": release_date,
        "release_country": release_country,
        "genres":genres,
        "user_score":score,
        "overview":overview,
        "director":directors,
        "writer": writers,
        "status": facts_clean["status"],
        "original_language":facts_clean["original_language"],
        "budget":facts_clean["budget"],
        "revenue":facts_clean["revenue"]
    }
```
