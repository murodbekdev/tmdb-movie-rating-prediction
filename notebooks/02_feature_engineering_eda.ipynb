# 1. Muhim kutubxonalarni import qilish 

```python
import pandas as pd
import numpy as np 

# Plotly.express 
import plotly.express as px
import plotly.graph_objects as go
```

# 2. Feature Engineered bo'lgan datasetni yuklash 

```python
df_fe = pd.read_csv("../data/feature_engineered/fe_dataset.csv")
df_fe.head(5)
```

# 3. release_year taqsimoti

Film qaysi yillarda ko'proq chiqqanini ko'rsatadi. 

```python
fig = px.histogram(
    df_fe,
    x = "release_year",
    title = "Chiqarilgan yil taqsimoti"
)

fig.show()
fig.write_image(
    "../results/figures/feature_engineered/release_year_dist.png"
)
```

Filmlar qaysi yillarda chiqqani ko'rsatadi. Asosan yillar oralig'ida miqdorini ko'rsatadi.


# 4. release_month taqsimoti 

```python
fig = px.histogram(
    df_fe,
    x="release_month",
    title = "Chiqarilgan oy taqsimoti"
)
fig.show()
fig.write_image(
    "../results/figures/feature_engineered/release_month_dist.png"
)
```

Bu yerda asosan qaysi oylarda ko'proq kinolar chiqishini ko'rishimiz mumkin.

# 5. movie_age taqsimoti 

```python
fig = px.histogram(
    df_fe,
    x="movie_age",
    title = "Film yoshi taqsimoti"
)
fig.show()
fig.write_image(
    "../results/figures/feature_engineered/movie_age_dist.png"
)
```

Bu yerda esa nechta kinolar necha yoshda ekanligi ko'rishimiz mumkin. 

# 5. movie_age bilan user_scoreni taqsimoti

```python
fig = px.scatter(
    df_fe,
    x="movie_age",
    y="user_score",
    trendline="ols",
    opacity=0.4,
    title="Kino Yoshi vs foydalanuvchi Bahosi"
)
fig.show()
fig.write_image(
    "../results/figures/feature_engineered/movi_age_vs_target_dist.png"
)
```

Film yoshi oshgani sari foydalanuvchi bahosi ozgina oshishga moyil.

# 6. Bog'liqlik Matriksi 

```python
corr = df_fe.select_dtypes(include=np.number).corr()
fig = px.imshow(
    corr, 
    text_auto = True,
    title = "Bog'liqlik Matriksi",
    width = 1000,
    height=750
)
fig.update_traces(
    hovertemplate=
    "X o'qi: %{x}<br>"
    "Y o'qi: %{y}<br>"
    "Bog'liqlik: %{z:.3f}<extra></extra>"
)
fig.show()
fig.write_image(
    "../results/figures/feature_engineered/target_corr_matrix.png"
)
```

Bu yerda X o'qi va Y o'qi va ularning bo'gliqligini ko'rishimiz mumkin.
